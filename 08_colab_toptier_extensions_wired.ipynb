```python
# 3. Load Dataloaders and Pre-trained Models
from src.data.cdf_reader import read_goes_directory, read_wind_directory
from src.data.preprocessor import Preprocessor
from src.data.dataloader import make_dataloaders
from src.model.storm_physnet import STORMPhysNet
from src.model.baselines import VanillaTransformer

print("Loading dataset into memory...")
goes = read_goes_directory("datasets/goes")
wind = read_wind_directory("datasets/omni")
raw  = goes.join(wind, how="inner")

pre_path = next(Path("checkpoints").rglob("preprocessor.pkl"), None)
pre = pickle.load(open(pre_path, "rb")) if pre_path else Preprocessor()
train_df, val_df, test_df = pre.fit_transform(raw)

_, _, test_loader = make_dataloaders(train_df, val_df, test_df, seq_len=72, batch_size=128, storm_weight=10.0, num_workers=0)
n_sw = test_loader.dataset.n_sw_features

def load_model(label):
    d = Path(f"checkpoints/{label}/seed_42")
    ckpt_cands = sorted(d.glob("*_best.pt")) + sorted(d.glob("*_best.zip"))
    if not ckpt_cands:
        raise FileNotFoundError(f"No checkpoint found for {label} in {d}")
    ckpt = ckpt_cands[0]
    if label == "storm_bz":
        model = STORMPhysNet(n_sw, 72, 128, 4, 2, 2, 64, 256, 64, 3, 0.1, "none", "transformer", "bz", False)
    else:
        model = VanillaTransformer(n_sw_features=n_sw, seq_len=72, n_horizons=3)
    state = torch.load(ckpt, map_location=device)
    if "state_dict" in state: state = state["state_dict"]
    model.load_state_dict(state, strict=False)
    return model.to(device)

print("Loading models...")
model_storm = load_model("storm_bz")
model_tf = load_model("transformer")
print("Models loaded successfully!")
```

## 1. Uncertainty Estimation (Monte Carlo Dropout)
By keeping dropout activated during inference (model.train()), we sample multiple paths through the network. The variance across these samples gives us a confidence interval (epistemic uncertainty).

```python
def predict_with_uncertainty(model, loader, num_samples=30):
    # Enable dropout during inference
    model.train()
    
    batch = next(iter(loader))
    x_sw   = torch.nan_to_num(batch["x_sw"].to(device), nan=0.0)
    x_flux = torch.nan_to_num(batch["x_flux"].to(device), nan=0.0)
    y_true = batch["y_flux"].numpy()[:, 1] # 6h true
    
    preds = []
    with torch.no_grad():
        for _ in range(num_samples):
            out = model(x_sw, x_flux, batch["y_persist"].to(device))
            yp = out["flux_pred"][:, 1].cpu().numpy() if isinstance(out, dict) else out[:, 1].cpu().numpy()
            preds.append(yp)
            
    preds = np.stack(preds)
    mean_pred = np.mean(preds, axis=0)
    std_pred = np.std(preds, axis=0)
    
    plt.figure(figsize=(10, 4), dpi=120)
    plt.plot(y_true[:80], label="True GOES Flux", color="black", lw=1.5)
    plt.plot(mean_pred[:80], label="STORM-PhysNet Mean", color="#E91E63", lw=1.5)
    plt.fill_between(range(80), mean_pred[:80] - 2*std_pred[:80], mean_pred[:80] + 2*std_pred[:80], 
                     alpha=0.3, color="#E91E63", label="95% Confidence Interval (MC Dropout)")
    plt.xlabel("Sample Index (Test Set)")
    plt.ylabel("Log Electron Flux")
    plt.title("Epistemic Uncertainty via Monte Carlo Dropout")
    plt.legend(frameon=False)
    plt.show()

predict_with_uncertainty(model_storm, test_loader)
```

## 2. Feature Importance (Integrated Gradients via Captum)
Calculates exactly which solar wind features (Bz, Density, Speed, etc) the model relied on to make its predictions.

```python
from captum.attr import IntegratedGradients

class ModelWrapper(nn.Module):
    def __init__(self, model):
        super().__init__()
        self.model = model
    def forward(self, x_sw):
        b, s, _ = x_sw.shape
        x_flux = torch.zeros(b, s, 1, device=device)
        out = self.model(x_sw, x_flux)
        return out["flux_pred"][:, 1] if isinstance(out, dict) else out[:, 1]

wrapped_model = ModelWrapper(model_storm)
ig = IntegratedGradients(wrapped_model)

batch = next(iter(test_loader))
x_sw = torch.nan_to_num(batch["x_sw"].to(device), nan=0.0)[:20] # Take 20 windows
baseline = torch.zeros_like(x_sw)

print("Calculating Integrated Gradients...")
attributions, delta = ig.attribute(x_sw, baseline, return_convergence_delta=True)
feat_importance = attributions.mean(dim=(0, 1)).detach().cpu().numpy()

# Names of your solar wind features (aligned with OMNI dataset columns)
sw_features = ["Speed", "Density", "Pressure", "Bt", "Bz", "By", "Ey", "Dst", "Kp", "AE", "AL", "AU", "Temp", "Beta", "LogFlux", "Other"][:n_sw]

plt.figure(figsize=(10, 4), dpi=120)
plt.bar(sw_features, feat_importance, color="#2196F3")
plt.xticks(rotation=45, ha='right')
plt.title("Global Feature Importance (Integrated Gradients) for 6h Forecast")
plt.ylabel("Average Attribution Score")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

## 3. Statistical Significance Testing
Proving that STORM-PhysNet's performance gain over the Vanilla Transformer is statistically significant.

```python
def run_significance_test(model_a, model_b, loader):
    model_a.eval()
    model_b.eval()
    
    sq_err_a, sq_err_b = [], []
    with torch.no_grad():
        for batch in loader:
            x_sw   = torch.nan_to_num(batch["x_sw"].to(device), nan=0.0)
            x_flux = torch.nan_to_num(batch["x_flux"].to(device), nan=0.0)
            yp     = batch["y_persist"].to(device)
            yt     = batch["y_flux"].numpy()[:, 1] # 6h true
            
            out_a = model_a(x_sw, x_flux, yp)
            out_b = model_b(x_sw, x_flux, yp)
            yp_a = out_a["flux_pred"][:, 1].cpu().numpy() if isinstance(out_a, dict) else out_a[:, 1].cpu().numpy()
            yp_b = out_b["flux_pred"][:, 1].cpu().numpy() if isinstance(out_b, dict) else out_b[:, 1].cpu().numpy()
            
            sq_err_a.extend((yt - yp_a)**2)
            sq_err_b.extend((yt - yp_b)**2)
            
    # Paired t-test
    # A = STORM, B = Baseline
    t_stat, p_value = stats.ttest_rel(sq_err_b, sq_err_a) 
    
    print(f"Paired t-test on 6h Squared Errors (Transformer vs STORM-BzGate):\n")
    print(f"T-statistic: {t_stat:.4f}")
    print(f"P-value:     {p_value:.4e}\n")
    if p_value < 0.05:
        print("\u2705 Result: The improvement of STORM-PhysNet over the Baseline is STATISTICALLY SIGNIFICANT (p < 0.05).")
    else:
        print("\u274c Result: The difference is not statistically significant.")

run_significance_test(model_storm, model_tf, test_loader)
```
